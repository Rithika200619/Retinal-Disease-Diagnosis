```python
def generate_split(master_df, selected_labels, seed, split_name, project_root):
    """
    Generate a multilabel stratified Train/Validation/Test split.

    Parameters
    ----------
    master_df : pandas.DataFrame
        Master dataset.

    selected_labels : list
        Disease label columns.

    seed : int
        Random seed.

    split_name : str
        Folder name to save the split.

    project_root : str
        Project root directory.

    Returns
    -------
    train_df, validation_df, test_df
    """

    print("=" * 60)
    print(f"Generating {split_name}")
    print("=" * 60)

    # ==========================================================
    # STEP 1 : Prepare Data
    # ==========================================================
    X = master_df[["ID", "New_Image_Name"]].copy()
    Y = master_df[selected_labels].values
    indices = np.arange(len(master_df))

    print("\nSTEP 1 : DATA PREPARATION")
    print("X Shape :", X.shape)
    print("Y Shape :", Y.shape)
    print("Indices :", len(indices))

    # ==========================================================
    # STEP 2 : Train / Temporary Split (70 / 30)
    # ==========================================================
    print("\nSTEP 2 : GENERATING TRAIN / TEMPORARY SPLIT")

    splitter = MultilabelStratifiedShuffleSplit(
        n_splits=1,
        test_size=0.30,
        random_state=seed
    )

    train_idx, temp_idx = next(splitter.split(indices, Y))

    train_df = master_df.iloc[train_idx].reset_index(drop=True)
    temp_df = master_df.iloc[temp_idx].reset_index(drop=True)

    print("\nTrain Images     :", len(train_df))
    print("Temporary Images :", len(temp_df))
    print("Total Images     :", len(train_df) + len(temp_df))

    # ==========================================================
    # STEP 3 : Validation / Test Split (15 / 15)
    # ==========================================================
    print("\nSTEP 3 : GENERATING VALIDATION / TEST SPLIT")

    temp_Y = temp_df[selected_labels].values
    temp_indices = np.arange(len(temp_df))

    splitter_temp = MultilabelStratifiedShuffleSplit(
        n_splits=1,
        test_size=0.50,
        random_state=seed
    )

    val_idx, test_idx = next(splitter_temp.split(temp_indices, temp_Y))

    validation_df = temp_df.iloc[val_idx].reset_index(drop=True)
    test_df = temp_df.iloc[test_idx].reset_index(drop=True)

    print("\nValidation Images :", len(validation_df))
    print("Test Images       :", len(test_df))

    print("\nFinal Split Summary")
    print("-" * 30)
    print("Training   :", len(train_df))
    print("Validation :", len(validation_df))
    print("Testing    :", len(test_df))
    print("Total      :", len(train_df) +
          len(validation_df) + len(test_df))

    # ==========================================================
    # STEP 4 : Disease Distribution
    # ==========================================================
    print("\n" + "=" * 60)
    print("STEP 4 : DISEASE DISTRIBUTION")
    print("=" * 60)

    distribution = []

    for disease in selected_labels:

        train_count = int(train_df[disease].sum())
        val_count = int(validation_df[disease].sum())
        test_count = int(test_df[disease].sum())

        total = train_count + val_count + test_count

        distribution.append([
            disease,
            train_count,
            val_count,
            test_count,
            total
        ])

    distribution_df = pd.DataFrame(
        distribution,
        columns=[
            "Disease",
            "Train",
            "Validation",
            "Test",
            "Total"
        ]
    )

    print(distribution_df)

    # ==========================================================
    # STEP 5 : Data Leakage Check
    # ==========================================================
    print("\n" + "=" * 60)
    print("STEP 5 : DATA LEAKAGE CHECK")
    print("=" * 60)

    train_ids = set(train_df["New_Image_Name"])
    validation_ids = set(validation_df["New_Image_Name"])
    test_ids = set(test_df["New_Image_Name"])

    train_val_overlap = train_ids.intersection(validation_ids)
    train_test_overlap = train_ids.intersection(test_ids)
    val_test_overlap = validation_ids.intersection(test_ids)

    print(f"Train ∩ Validation : {len(train_val_overlap)}")
    print(f"Train ∩ Test       : {len(train_test_overlap)}")
    print(f"Validation ∩ Test  : {len(val_test_overlap)}")

    if (len(train_val_overlap) == 0 and
        len(train_test_overlap) == 0 and
        len(val_test_overlap) == 0):

        print("\n✅ No data leakage detected.")

    else:

        print("\n❌ Data leakage detected.")

    # ==========================================================
    # STEP 6 : Save Files
    # ==========================================================
    print("\n" + "=" * 60)
    print("STEP 6 : SAVING SPLIT FILES")
    print("=" * 60)

    split_folder = os.path.join(
        project_root,
        "splits",
        split_name
    )

    os.makedirs(split_folder, exist_ok=True)

    train_df.to_csv(
        os.path.join(split_folder, "train.csv"),
        index=False
    )

    validation_df.to_csv(
        os.path.join(split_folder, "validation.csv"),
        index=False
    )

    test_df.to_csv(
        os.path.join(split_folder, "test.csv"),
        index=False
    )

    print("✓ train.csv saved")
    print("✓ validation.csv saved")
    print("✓ test.csv saved")

    # ==========================================================
    # STEP 7 : Save Split Information
    # ==========================================================
    info_path = os.path.join(split_folder, "split_info.txt")

    with open(info_path, "w") as f:

        f.write(f"Split Name : {split_name}\n")
        f.write(f"Random Seed : {seed}\n\n")

        f.write(f"Training Images : {len(train_df)}\n")
        f.write(f"Validation Images : {len(validation_df)}\n")
        f.write(f"Testing Images : {len(test_df)}\n\n")

        f.write("Disease Distribution\n")
        f.write("====================\n")
        f.write(distribution_df.to_string(index=False))

    print("✓ split_info.txt saved")

    # ==========================================================
    # STEP 8 : Verify Saved Files
    # ==========================================================
    print("\n" + "=" * 60)
    print("STEP 8 : VERIFY SAVED FILES")
    print("=" * 60)

    train_check = pd.read_csv(
        os.path.join(split_folder, "train.csv")
    )

    validation_check = pd.read_csv(
        os.path.join(split_folder, "validation.csv")
    )

    test_check = pd.read_csv(
        os.path.join(split_folder, "test.csv")
    )

    print("Train Shape      :", train_check.shape)
    print("Validation Shape :", validation_check.shape)
    print("Test Shape       :", test_check.shape)

    print("\n✓ Saved files verified successfully.")

    return train_df, validation_df, test_df
```

```python
def compare_splits(df1, df2, name1, name2):
    s1 = set(df1["New_Image_Name"])
    s2 = set(df2["New_Image_Name"])

    common = len(s1 & s2)
    only_first = len(s1 - s2)
    only_second = len(s2 - s1)

    print(f"\n{name1} vs {name2}")
    print("-" * 35)
    print("Common images :", common)
    print(f"Only in {name1} :", only_first)
    print(f"Only in {name2} :", only_second)

compare_splits(train_df_split2, train_df_split3, "Split2", "Split3")
compare_splits(train_df_split2, train_df_split4, "Split2", "Split4")
compare_splits(train_df_split2, train_df_split5, "Split2", "Split5")
```
